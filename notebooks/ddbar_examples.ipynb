```python
X = ComplexChart(2, names=("z","w")); X
```

```python
(z,w) = X.coordinates()
```

### Cone metrics

```python
beta = var('beta', domain='real', latex_name=r'\beta')
```

```python
beta
```

```python
ddbar(abs(z)**(2*beta))
```

```python
f = log (1+ abs(z)**(2*beta)); f
```

```python
ddbar(f)
```

### Eguchi Hanson metric

```python
s = abs(z)^2 + abs(w)^2; s
```

```python
u = sqrt(s^2 +1) + log(s/(sqrt(s^2+1)+1)); u
```

```python
EH_metric = ddbar(u); EH_metric
```

```python
det(EH_metric)
```

### Fubini-Study metric

```python
FS_metric = ddbar(log(1+s)); FS_metric
```

```python
FS_metric.det()
```

```python
FS_metric.coefficient_matrix()
```

### Calabi-Yau metric on $\mathbb{C}^2$ with cone angle $\pi$ along the conic $zw=1$

```python
phi = sqrt(s + 2*abs(z*w-1) +2); phi
```

```python
det(ddbar(phi))
```

```python
asymptotic_cone = ddbar(abs(z)+abs(w))
```

```python
asymptotic_cone
```
